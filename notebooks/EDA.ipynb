# Análise Exploratória — Sprint 5
Exploração básica do conjunto de dados de anúncios de automóveis.

```python
from pathlib import Path
import pandas as pd
import plotly.express as px

csv_file = Path("vehicles_us.csv")
if not csv_file.exists():
    csv_file = Path("..") / "vehicles_us.csv"
car_data = pd.read_csv(csv_file)
car_data.head()
```

## 1. Dimensões, colunas e valores em falta

```python
print("Dimensões:", car_data.shape)
print("Valores em falta por coluna:")
print(car_data.isna().sum())
car_data[["odometer", "price"]].describe()
```

## 2. Histograma de quilometragem

```python
fig_hist = px.histogram(car_data, x="odometer", nbins=50, title="Distribuição da quilometragem")
fig_hist.show()
```

## 3. Dispersão: quilometragem e preço

```python
fig_scatter = px.scatter(car_data.dropna(subset=["odometer", "price"]), x="odometer", y="price", title="Quilometragem vs. preço")
fig_scatter.show()
```

## Conclusão
O notebook permite inspecionar valores em falta, a distribuição de quilometragem e a relação entre uso e preço dos veículos. O foco do Sprint 5 é disponibilizar estas visualizações numa aplicação pública.
